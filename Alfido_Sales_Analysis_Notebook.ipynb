# Alfido Tech - Sales Performance Analysis

## Import Libraries

```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Load Dataset

```

df = pd.read_csv('alfido_sales_data.csv')
df.head()
```

## Data Cleaning

```

df.dropna(inplace=True)
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
```

## KPI Calculations

```

total_revenue = df['Revenue'].sum()
total_profit = df['Profit'].sum()
profit_margin = (total_profit / total_revenue) * 100
avg_order_value = df['Revenue'].mean()

print(total_revenue, total_profit, profit_margin, avg_order_value)
```

## Monthly Revenue Trend

```

df['Month'] = df['OrderDate'].dt.to_period('M').astype(str)
monthly_sales = df.groupby('Month')['Revenue'].sum()

monthly_sales.plot(figsize=(10,5))
plt.title('Monthly Revenue Trend')
plt.show()
```

## Top Products

```

top_products = df.groupby('Product')['Revenue'].sum().sort_values(ascending=False)

top_products.plot(kind='bar')
plt.title('Top Performing Products')
plt.show()
```

## Regional Sales

```

regional_sales = df.groupby('Region')['Revenue'].sum()

regional_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Regional Revenue Distribution')
plt.ylabel('')
plt.show()
```
